NEA ANALYSIS 

The project is to code a Rubik’s Cube solver that will allow a non-expert user to photograph a scrambled cube and will then output a visual guide for the user to  solve the cube. 

In a ‘virtual’ mode the programme will allow the user to scramble and solve a digital cube on screen. 

I researched the mathematics and linear algebra behind 3D engines to learn how to project a three-dimensional Rubik’s Cube on a two-dimensional screen and rotate the cube. 

I researched the appropriateness of several computer algorithms to solve a Rubik’s Cube, including Kociemba’s Algorithm, Korf’s Algorithm and CFOP.

I also researched the theory of how to solve the Rubik’s Cube, what constitutes a possible configuration for a cube, the number of moves required to solve a cube and the computational demands of different solving algorithms. 

I also looked into different libraries for computer vision and colour detection. 

The Rubik’s Cube is a three-dimensional puzzle invented by Erno Rubik in 1974. The Cube’s six faces each have nine coloured ‘facelets’ and the facelets can be rotated into different positions on the cube by turning the cube’s faces in any of six directions. The objective is to ‘solve’ the cube by using sequences of turns to reposition all the facelets so that each whole face of the Rubik’s Cube is a single colour. The solution to the puzzle depends always on the configuration of a scrambled cube. The total number of possible configurations is roughly 43 quintillion.

The objectives of the project are: 

3D ENGINE
    1.    Display a representation of a perspective image of a three-dimensional Rubik’s Cube on screen.
    2.    The image must include coloured facelets.
    3.    Allow the user to rotate the image about all three axes using keyboard inputs.
    4.    Allow the user to input a turn to manipulate the cube’s facelets.
    5.    Display smooth animations of face turns and cube rotations so the user can follow along easily.

SOLVING ALGORITHM 
    1.    Generate a solution for a scrambled Rubik’s Cube using Kociemba’s Algorithm, within reasonable time — ideally, under two seconds.
    2.    Generate and store ‘move tables’ and ‘pruning tables’ used in Kociemba’s Algorithm.

CUBE CAPTURE
    1.    Accurately capture the configuration of a Rubik’s Cube by sequentially photographing its faces using a webcam.
    2.    Adapt to different colour schemes of different Rubik’s Cubes.
    3.    Allow for different lighting conditions and webcam quality.
    4.    Give the user the opportunity to verify and correct a scanned cube if the colour recognition fails.
    5.    Verify that a captured configuration is solvable.

GENERAL 
    1.    Allow generation of random valid scrambles for a Rubik’s Cube.
    2.    Allow users to input standard Rubik’s Cube notation to describe a configuration if they choose to do so, as an alternative to a random scramble. The input must be checked for standard grammar.
    3.    Show the solution to the user in groups of three turns at a time, with the user able to replay or undo turns.
    4.    The solver needs to be intuitive and easy to use for someone with no prior ability to successfully solve a Rubik’s Cube.
